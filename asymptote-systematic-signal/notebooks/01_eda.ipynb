# 01 — Exploratory Data Analysis (SYNTHETIC DATA)

All data here is synthetic (seed 42). Tickers are prefixed `SYN_` so nothing can be mistaken for live prices.

```python
import sys
sys.path.insert(0, '../src')
import pandas as pd, numpy as np
pd.set_option('display.max_columns', 20)
from data_pipeline import generate_synthetic_prices, generate_synthetic_kpis
prices = generate_synthetic_prices()
kpis = generate_synthetic_kpis(prices)
print(prices.shape)
print(prices.dtypes)
```

```python
prices.head(10)
```

```python
prices.describe().T
```

```python
print('Regime counts:')
print(prices['regime'].value_counts())
print()
print('Daily returns by regime (SYN_EQ):')
print(prices.assign(ret=prices['SYN_EQ'].pct_change()).groupby('regime')['ret'].agg(['mean','std','count']))
```

```python
kpis.tail(12)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(prices['date'], prices['SYN_EQ'], lw=0.8)
ax.set_title('SYN_EQ synthetic price (2019-2025)')
plt.show()
```
